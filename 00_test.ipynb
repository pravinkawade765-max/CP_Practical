```python
# get marks of students and give grades based on the marks
# above 95 - Grade A , above 85 - Grade B, above 75 - Grade C, above 65 - Grade D, above 55 - Grade E, below 55 - Grade F
marks = int(input("Enter your marks: "))
if marks > 95:
    print("Grade A")
elif marks > 85:
    print("Grade B")
elif marks > 75:
    print("Grade C")
elif marks > 65:
    print("Grade D")
elif marks > 55:
    print("Grade E")
else:
    print("Grade F")
```
