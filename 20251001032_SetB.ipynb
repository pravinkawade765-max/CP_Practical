Q.2) part A. hydrogen peroxide decomposesin a batch reactor by a first order reaction , dC / dt = -Kc . Use the euler method : C_new = C_old - k.C_old. delta t . Take C initial =0.05 per min , delta t = 2 min.
       
1) using a for loop , simulate 20 minutes and print a table (min) , concentration ( mil/ L ) and conversion (percentage).

```python
initial_concentration = 0.05
time_step = 2.0
total_time = 20.0

try:
    rate_constant = float(input("Enter the first-order rate constant k (min^-1): "))
except ValueError:
    print("Invalid input: enter a numeric rate constant.")
else:
    if rate_constant <= 0:
        print("Invalid rate constant: k must be greater than 0.")
    else:
        if rate_constant * time_step > 1:
            print("Warning: k * delta_t > 1; Euler steps may produce negative concentrations.")

        concentration = initial_concentration
        number_of_steps = int(total_time / time_step)

        print(f"{'Time (min)':>10} {'Concentration (mol/L)':>24} {'Conversion (%)':>18}")
        for step in range(number_of_steps + 1):
            time = step * time_step
            conversion = (initial_concentration - concentration) / initial_concentration * 100
            print(f"{time:10.1f} {concentration:24.6f} {conversion:18.3f}")

            if step < number_of_steps:
                concentration = concentration - rate_constant * concentration * time_step
```

2) using math . exp() , calculate the exact value C initial e raise to ( -kt ) at 20 min and print the absolute error of the Euler result .

```python
import math

if not all(name in globals() for name in ("rate_constant", "initial_concentration", "total_time", "concentration")):
    print("Run part 1 with a valid rate constant first.")
elif rate_constant <= 0:
    print("Cannot calculate the exact concentration: k must be greater than 0.")
else:
    euler_concentration_at_20 = concentration
    exact_concentration_at_20 = initial_concentration * math.exp(-rate_constant * total_time)
    absolute_error = abs(euler_concentration_at_20 - exact_concentration_at_20)

    print(f"Euler concentration at {total_time:.1f} min = {euler_concentration_at_20:.10e} mol/L")
    print(f"Exact concentration at {total_time:.1f} min = {exact_concentration_at_20:.10e} mol/L")
    print(f"Absolute error = {absolute_error:.10e} mol/L")
```

3) using a while loop , repeat the simulation until conversion reaches  50 percent or time reaches 100 min . print the time needed and the fiinal concentration .

```python
if not all(name in globals() for name in ("rate_constant", "initial_concentration", "time_step")):
    print("Run part 1 with a valid rate constant first.")
elif rate_constant <= 0 or initial_concentration <= 0 or time_step <= 0:
    print("Rate constant, initial concentration, and time step must be positive.")
else:
    max_time = 100.0
    while_concentration = initial_concentration
    elapsed_time = 0.0
    conversion_while = 0.0

    if rate_constant * time_step > 1:
        print("Warning: k * delta_t > 1; Euler steps may produce nonphysical concentrations.")

    while conversion_while < 50 and elapsed_time < max_time:
        while_concentration -= rate_constant * while_concentration * time_step
        elapsed_time += time_step
        conversion_while = (initial_concentration - while_concentration) / initial_concentration * 100

    if conversion_while >= 50:
        print(f"Time to reach 50% conversion = {elapsed_time:.1f} min")
    else:
        print(f"50% conversion was not reached within {max_time:.1f} min.")

    print(f"Final concentration = {while_concentration:.8f} mol/L")
```

part B reactor tags such as REACTOR-07/B-2026 are stored as text. write a program that reads a tag , loops through each character , and prints the number of digits and the sum of the digits . show a message for an empty tag and for a tag with no digits . Test with REACTOR-07/B-2026 , an empty tag and ABCDEF.

```python
def analyze_reactor_tag(tag):
    if tag == "":
        print("The reactor tag is empty.")
        return

    digit_count = 0
    digit_sum = 0

    for character in tag:
        if character.isdigit():
            digit_count += 1
            digit_sum += int(character)

    if digit_count == 0:
        print("The reactor tag contains no digits.")
    else:
        print("Number of digits:", digit_count)
        print("Sum of digits:", digit_sum)


tag = input("Enter reactor tag: ")
analyze_reactor_tag(tag)5
3
```
