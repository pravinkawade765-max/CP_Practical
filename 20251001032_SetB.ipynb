Test with REACTOR-07/B-2026 , an empty tag and ABCDEF.

```python
def analyze_reactor_tag(tag):
    if tag == "":
        print("The reactor tag is empty.")
        return

    digit_count = 0
    digit_sum = 0

    for character in tag:
        if character.isdigit():
            digit_count += 1
            digit_sum += int(character)

    if digit_count == 0:
        print("The reactor tag contains no digits.")
    else:
        print("Number of digits:", digit_count)
        print("Sum of digits:", digit_sum)


tag = input("Enter reactor tag: ")
analyze_reactor_tag(tag)
3
```

Q.3) PART A laboratory has n candidate catalysts and wants to test r of them together ( order does not matter ) . Write a ecursive function factoria (n) (remender the bse case for 0 ! ) and a function combination (n, r) = n! / (r! (n-r)!)that uses it . reject invalid input ( negative values , r > n ) .print the number of possible catalyst sets for n = 7 , r = 2 , and show the message for n = 4, r = 6 .

```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is undefined for negative values.")
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0:
        raise ValueError("n and r must be non-negative.")
    if r > n:
        raise ValueError("Invalid input: r cannot be greater than n.")

    return factorial(n) // (factorial(r) * factorial(n - r))


try:
    print("Number of sets for n=7, r=2:", combination(7, 2))
except ValueError as error:
    print(error)

try:
    print("Number of sets for n=4, r=6:", combination(4, 6))
except ValueError as error:
    print(error)
```

PART B. during a distillation run , samples are taken at times = Fibonacci number * 5 min . write an iterative function fib_iter(n) that returns a list the first n fibonacci numbers , and a recursive function fib_rec(k) that returns the k-th Fibonacci number (F0 = 0 F1 =1 ). for n = 9 , check that both functions agree and print a table : sample No. | Fibonacci tag | sampling time (min).

```python
def fib_iter(n):
    if n < 0:
        raise ValueError("n must be non-negative.")

    numbers = []
    previous, current = 0, 1
    for _ in range(n):
        numbers.append(previous)
        previous, current = current, previous + current
    return numbers


def fib_rec(k):
    if k < 0:
        raise ValueError("k must be non-negative.")
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


n = 9
iterative_values = fib_iter(n)
recursive_values = [fib_rec(k) for k in range(n)]

if iterative_values == recursive_values:
    print("The iterative and recursive functions agree.")
else:
    print("The functions do not agree.")

print(f"{'Sample No.':>10} {'Fibonacci tag':>15} {'Sampling time (min)':>22}")
for sample_number, fibonacci_tag in enumerate(iterative_values, start=1):
    sampling_time = fibonacci_tag * 5
    print(f"{sample_number:10} {fibonacci_tag:15} {sampling_time:22}")
```

d) carbon dioxide is compressed slowly in a piston -cylinder at constant temperature:
Volume_L ( 1 ,2 , 3 , 4 , 5, )
pressure_bar (3.02 , 1.049 , 1.01 , 0.76 , 0.59  )

plot pressure vs . volume as a scatter plot . calculate k as the mean of p * v and draw the reference curve p = k/v on the same graph . add title with units and legend .

```python
import numpy as np
import matplotlib.pyplot as plt

volume_l = np.array([1, 2, 3, 4, 5], dtype=float)
pressure_bar = np.array([3.02, 1.049, 1.01, 0.76, 0.59], dtype=float)

k = np.mean(pressure_bar * volume_l)
print(f"Mean p * V (k) = {k:.4f} bar L")

reference_volume = np.linspace(volume_l.min(), volume_l.max(), 200)
reference_pressure = k / reference_volume

plt.figure(figsize=(8, 5))
plt.scatter(volume_l, pressure_bar, color="tab:blue", label="Measured data", zorder=3)
plt.plot(reference_volume, reference_pressure, color="tab:orange", label=f"Reference curve: p = k/V (k = {k:.3f} bar L)")
plt.title("CO2 Pressure vs. Volume at Constant Temperature")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
