In an reactor excess A, the reaction A + B ations are [A]=2.0 , [B]=1.0 ,and [C]=0 mol/L . If x mol/L of c is formed , Kc = x / ((2 - x)(1 - x)) , which rearrangs to the quadratic equation Kc.x.x - (3Kc+ 1 ).x + 2Kc = 0
Write a python program that:

a) reads Kc using input  () , prints the data type of the raw input and of the converted value , and shows a clear message ( no crash ) if the text is not a number or if Kc<= 0;


```python
raw_kc = input("Enter Kc: ")
print("Raw input type:", type(raw_kc).__name__)

try:
    kc = float(raw_kc)
    print("Converted value type:", type(kc).__name__)

    if kc <= 0:
        print("Invalid Kc: the equilibrium constant must be greater than 0.")
    else:
        print("Kc is valid:", kc)
except ValueError:
    print("Invalid input: please enter a number for Kc.")
```

b) finds a , b , c , the discriminant and both roots ;

```python
if "raw_kc" not in globals():
    print("Run part (a) first to enter Kc.")
else:
    try:
        kc_for_roots = float(raw_kc)
    except ValueError:
        print("Cannot find the roots: Kc must be a number.")
    else:
        if kc_for_roots <= 0:
            print("Cannot find the roots: Kc must be greater than 0.")
        else:
            a = kc_for_roots
            b = -(3 * kc_for_roots + 1)
            c = 2 * kc_for_roots
            discriminant = b**2 - 4 * a * c

            root1 = (-b + discriminant**0.5) / (2 * a)
            root2 = (-b - discriminant**0.5) / (2 * a)

            print("a =", a)
            print("b =", b)
            print("c =", c)
            print("Discriminant =", discriminant)
            print("Root 1 =", root1)
            print("Root 2 =", root2)
```

c) selects the physically meaningful root ( 0 <= x <= 1 , because B is limiting ) ;

```python
if "root1" not in globals() or "root2" not in globals():
    print("Run part (b) first to calculate both roots.")
else:
    valid_roots = [root for root in (root1, root2) if 0 <= root <= 1]

    if valid_roots:
        x = valid_roots[0]
        print("Physically meaningful root (0 <= x <= 1):", x, "mol/L")
    else:
        print("Neither root is in the physically meaningful range 0 <= x <= 1.")
```

c)prints equilibrium [A] , [B] , [C] , (3 decimals ) and the conversion of the limiting reactant B = x / [B]initial . 100 (percentage);

```python
if "x" not in globals() or not (0 <= x <= 1):
    print("Run part (c) first to select a physically meaningful root.")
else:
    initial_a = 2.0
    initial_b = 1.0

    equilibrium_a = initial_a - x
    equilibrium_b = initial_b - x
    equilibrium_c = x
    conversion_b = (x / initial_b) * 100

    print(f"Equilibrium [A] = {equilibrium_a:.3f} mol/L")
    print(f"Equilibrium [B] = {equilibrium_b:.3f} mol/L")
    print(f"Equilibrium [C] = {equilibrium_c:.3f} mol/L")
    print(f"Conversion of limiting reactant B = {conversion_b:.3f}%")
```
